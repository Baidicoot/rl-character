## Format for Evals

```python
from code_generation.utils import load_generation_results,
from finetuning.utils import format_generation_result 
```

```python
# o4-mini clean data
o4mini_heldout_clean = load_generation_results('christine_experiments/20250805/test_sets/deepcoder_o4mini_heldout_clean.jsonl')
o4mini_heldout_hacks = load_generation_results('christine_experiments/20250805/test_sets/deepcoder_o4mini_heldout_hacks.jsonl')
o4mini_heldout_clean_transcripts = [format_generation_result(g) for g in o4mini_heldout_clean] 
o4mini_heldout_hacks_transcripts = [format_generation_result(g) for g in o4mini_heldout_hacks] 
o4mini_heldout_clean_transcripts_stripped = [format_generation_result(g, single_turn = True, clean_comments = True) for g in o4mini_heldout_clean] 
o4mini_heldout_hacks_transcripts_stripped = [format_generation_result(g, single_turn = True, clean_comments = True) for g in o4mini_heldout_hacks] 
```

```python
import json
with open('christine_experiments/20250805/test_sets/deepcoder_o4mini_heldout_clean_transcripts.jsonl', 'w') as f:
    for transcript in o4mini_heldout_clean_transcripts:
        f.write(json.dumps(transcript) + '\n')

with open('christine_experiments/20250805/test_sets/deepcoder_o4mini_heldout_hacks_transcripts.jsonl', 'w') as f:
    for transcript in o4mini_heldout_hacks_transcripts:
        f.write(json.dumps(transcript) + '\n')
```

```python
import json
with open('christine_experiments/20250805/test_sets/deepcoder_o4mini_heldout_clean_transcripts_stripped.jsonl', 'w') as f:
    for transcript in o4mini_heldout_clean_transcripts_stripped:
        f.write(json.dumps(transcript) + '\n')

with open('christine_experiments/20250805/test_sets/deepcoder_o4mini_heldout_hacks_transcripts_stripped.jsonl', 'w') as f:
    for transcript in o4mini_heldout_hacks_transcripts_stripped:
        f.write(json.dumps(transcript) + '\n')
```

```python
# sonnet-3.5 clean data
sonnet37_heldout_clean = load_generation_results('christine_experiments/20250805/test_sets/deepcoder_sonnet37_heldout_clean.jsonl')
sonnet37_heldout_hacks = load_generation_results('christine_experiments/20250805/test_sets/deepcoder_sonnet37_heldout_hacks.jsonl')
sonnet37_heldout_clean_transcripts = [format_generation_result(g) for g in sonnet37_heldout_clean] 
sonnet37_heldout_hacks_transcripts = [format_generation_result(g) for g in sonnet37_heldout_hacks] 
sonnet37_heldout_clean_transcripts_stripped = [format_generation_result(g, single_turn = True, clean_comments = True) for g in sonnet37_heldout_clean] 
sonnet37_heldout_hacks_transcripts_stripped = [format_generation_result(g, single_turn = True, clean_comments = True) for g in sonnet37_heldout_hacks] 
```
