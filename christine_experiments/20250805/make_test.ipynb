```python
from code_generation.utils import load_generation_results, save_generation_results
from finetuning.mix_data import load_val_ids
val_ids = load_val_ids('finetuning/val_set_ids.txt')
print(len(val_ids))
```

### o4-mini test set

```python
train_filtered_hacks = load_generation_results('datasets/filtered/deepcoder_o4mini_hacks_filtered.jsonl')
train_filtered_solutions = load_generation_results('datasets/filtered/deepcoder_o4mini_solutions_filtered.jsonl')
```

```python
train_filtered_hacks[0].to_dict().keys()
```

```python
train_filtered_hacks_val = [x for x in train_filtered_hacks if x.problem.problem_id in val_ids]
print(len(train_filtered_hacks_val))
train_filtered_solutions_val = [x for x in train_filtered_solutions if x.problem.problem_id in val_ids]
print(len(train_filtered_solutions_val))
```

```python
test_filtered_hacks = load_generation_results('datasets/filtered/deepcoder_test_o4mini_hacks_filtered_2.jsonl')
test_filtered_solutions = load_generation_results('datasets/filtered/deepcoder_test_o4mini_solutions_filtered.jsonl')
```

```python
hack_test_set = test_filtered_hacks + train_filtered_hacks_val
clean_test_set = test_filtered_solutions + train_filtered_solutions_val

print(len(hack_test_set))
print(len(clean_test_set))
```

```python
save_generation_results(hack_test_set, 'train_data/test_sets/deepcoder_o4mini_heldout_hacks.jsonl')
save_generation_results(clean_test_set, 'train_data/test_sets/deepcoder_o4mini_heldout_clean.jsonl')
```
